### Hough Transform

> This notebook demonstrates the Hough Transform technique for detecting lines in an image. <br/>
The process involves reading an image, converting it to grayscale, applying Canny edge detection, and then using the Hough Transform to identify and draw lines based on the detected edges. <br/>
 The results are visualized using matplotlib.

import required libraries

- cv2 - for image processing
- math - for mathematical operations
- numpy - for numerical operations
- seaborn - for data visualization
- matplotlib.pyplot - for plotting images and graphs

```python
import cv2
import math
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
```

set up plotting parameters for better visualization of images and graphs. The figure size is set to 15x10 inches and the style is set to 'whitegrid' for a clean look.

```python
plt.rcParams['figure.figsize'] = [15, 10]
sns.set_style('whitegrid')
```

read image and convert it to grayscale, then apply Canny edge detection to find edges in the image. The original image and the edge-detected image are displayed side by side using matplotlib.

```python
img = cv2.imread('../images/yield.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
edges = cv2.Canny(gray, threshold1=100, threshold2=550)

plt.subplot(121), plt.imshow(img), plt.grid(False)
plt.subplot(122), plt.imshow(edges, cmap='gray'), plt.grid(False)
```

initialize HoughLines and draw lines on a blank image based on the detected edges. The original edge-detected image and the image with Hough lines are displayed side by side using matplotlib.

```python
lines = cv2.HoughLines(edges,
                       rho=1, # Distance resolution in pixels
                       theta=2*np.pi/180, # Angle resolution in radians
                       threshold=50) # Min number of votes
hough = np.zeros_like(edges)


for i in range(0, len(lines[0:3])):
    rho = lines[i][0][0]
    theta = lines[i][0][1]
    a = math.cos(theta)
    b = math.sin(theta)
    x0 = a * rho
    y0 = b * rho
    pt1 = (int(x0 + 1000 * (-b)), int(y0 + 1000 * a))
    pt2 = (int(x0 - 1000 * (-b)), int(y0 - 1000 * a))
    cv2.line(hough, pt1, pt2, 255, 1, cv2.LINE_AA)

plt.subplot(121), plt.imshow(edges, cmap='gray'), plt.grid(False)
plt.subplot(122), plt.imshow(hough, cmap='gray'), plt.grid(False)
```
