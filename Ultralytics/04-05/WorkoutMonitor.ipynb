# Workout Solution
- to track the left hand of a person doing a workout, we can use the `AIGym` solution from Ultralytics. This solution uses a pose estimation model to detect keypoints on the human body, allowing us to monitor specific movements during exercise.

import required libraries
- cv2 - OpenCV library for video processing
- ultralytics - Ultralytics library for pose estimation and workout monitoring

```python
import cv2
from ultralytics import solutions
```

```python
cap = cv2.VideoCapture("../media/pose_video.mov")
w, h, fps = (
    int(cap.get(x))
    for x in (cv2.CAP_PROP_FRAME_WIDTH,
              cv2.CAP_PROP_FRAME_HEIGHT,
              cv2.CAP_PROP_FPS))
video_writer = (
    cv2.VideoWriter(
        "result.avi",
        cv2.VideoWriter_fourcc(*"mp4v"),
        fps, (w, h)))
```

Workout Module Initialization

```python
solution = solutions.AIGym(
    model="../models/yolov8n-pose.pt",
    kpts=[6, 8, 10], # monitor only left hand
    line_width=5,
)
```

Process video

```python
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        break
    results = solution(im0)
    video_writer.write(results.plot_im)
cap.release()
video_writer.release()
```
