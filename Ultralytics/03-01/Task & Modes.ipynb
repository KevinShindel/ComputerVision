# Tasks

- tasks are the main functions of the model, such as detection, segmentation, classification, etc.
- each task has its own model architecture and weights, which are trained for that specific task.

```python
from ultralytics import YOLO
```

Load segmentation model for Segmentation task

```python
model = YOLO("../models/yolo11n-seg.pt")  # load an official model
```

# Modes
- modes are the different ways to use the model, such as predict, train, val, export, etc.
- each mode has its own set of parameters and options, which can be customized for different use cases.

```python
source = "../media/bottles_video.mov"

# Predict with the model
results = model.predict(
    source=source,
    # show=True
    save=True,
    )

# Access the results
# for result in results:
#     xywh = result.boxes.xywh  # center-x, center-y, width, height
#     xywhn = result.boxes.xywhn  # normalized
#     xyxy = result.boxes.xyxy  # top-left-x, top-left-y, bottom-right-x, bottom-right-y
#     xyxyn = result.boxes.xyxyn  # normalized
#     names = [result.names[cls.item()] for cls in result.boxes.cls.int()]  # class name of each box
#     confs = result.boxes.conf  # confidence score of each box

# Predict with the model
results = model.track(
    source=source,
    show=True
    # save=True,
    )
```
