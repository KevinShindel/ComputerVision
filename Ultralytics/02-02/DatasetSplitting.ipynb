# Dataset Splitting

- unzip the dataset
- split the dataset into train, validation and test sets

import required libraries
- zipfile: to unzip the dataset
- splitfolders: to split the dataset into train, validation and test sets

```python
from zipfile import ZipFile

import splitfolders
```

function to unzip the dataset

```python
def unzip_file(zip_path, extract_to):
    """Extracts a ZIP file."""
    with ZipFile(zip_path,
                 'r') as zip_ref:
        zip_ref.extractall(extract_to)
```

run the function to unzip the dataset

```python
unzip_file("export.zip",
           "export")
```

split the dataset into train, validation and test sets

```python
splitfolders.ratio("export",
                   "dataset",
                   ratio=(0.7, 0.15, 0.15),
                   seed=42)
```
