# Human Pose Estimation with YOLOv11n-Pose

import required libraries

```python
from ultralytics import YOLO
```

init the default pose model

```python
model = YOLO("../models/yolo11n-pose.pt")
```

inference on a video file

```python
model.predict(
    source="../media/pose_video.mov",
    line_width=7,
    conf=0.5,
    save=True,
    verbose=False,
)
```
