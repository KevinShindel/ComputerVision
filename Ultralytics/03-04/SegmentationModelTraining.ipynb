# Train a segmentation model using the auto-annotated dataset

Import the YOLO class from the ultralytics library

```python
from ultralytics import YOLO
```

Create a YOLO model instance using the best weights from the previous training run

```python
model = YOLO("../models/yolov8n-seg.pt") # load a pretrained segmentation model
```

Train the model using the auto-annotated dataset

```python
model.train(
    data="data.yaml", # use YAML configuration for datasets and labels
    batch=16,         # batch-size
    workers=1,        # number of workers ( aka
    epochs=15,       # number of max epochs
)
```

Evaluate the trained model on the validation dataset

```python
trained_model = YOLO("../runs/segment/train/weights/best.pt")

trained_model.predict(source="../03-02/video.mov",
              save=True,
              line_width=2,
              show_boxes=False)
```
