# Supported Models

import libraries

- **SAM** Segmentation And Masking model
- **FastSAM** Segmentation And Masking model

```python
from ultralytics import SAM, FastSAM
```

create a SAM model and run inference on an image

```python
# Create a FastSAM model
model = FastSAM("../models/FastSAM-s.pt")  # or FastSAM-x.pt

# Run inference on an image
everything_results = model.predict(
    source="../media/image.png",
    device="cpu",
    retina_masks=True,
    imgsz=1024,
    conf=0.4,
    iou=0.9,
    save=True
)
```

```python
# Load a model
model = SAM("sam2.1_b.pt")

# Run inference
model.predict(
    source="../media/image.png", save=True)
```
