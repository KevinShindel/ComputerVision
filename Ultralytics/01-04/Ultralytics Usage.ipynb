# Basic Ultralytics YOLO using

import required libraries
- **cv2** - for image processing
- **ultralytics** - for YOLO model
- **matplotlib.pyplot** - for displaying images

```python
import cv2
from ultralytics import YOLO
import matplotlib.pyplot as plt
```

## Detect Poses
- used for detecting human poses in images or videos.

```python
model_version = '../models/yolo11n-pose.pt'
model = YOLO(model_version)
```

read an image and display it using matplotlib

```python
img = cv2.imread('../media/pose_image.png', cv2.IMREAD_COLOR_RGB)

plt.imshow(img)
```

use inference to detect poses in the image and save the results

```python
results = model.predict(source="../media/pose_image.png",
                        save=True)
```

inference results can be accessed using the `results` object, which contains information about the detected poses, including keypoints and bounding boxes.

```python
for result in results:
    print(result.keypoints)
```

show bounding boxes of detected poses

```python
for result in results:
    print(result.boxes.xyxy)
```

use segmentation model to detect objects in an image and save the results
