# YOLO Model Export

import required libraries

```python
from ultralytics import YOLO
```

init YOLO model with the path to the trained weights file

```python
model = YOLO("yolo11n-pose.pt")
```

export the model to TorchScript format with specified parameters:
- **format**: "torchscript" - specifies the export format as TorchScript
- **batch**: 2 - specifies the batch size for the exported model
- **imgsz**: 416 - specifies the input image size for the exported model

```python
model.export(
    format="torchscript",
    # simplify=True,
    batch=2,
    imgsz=416,
)
```
