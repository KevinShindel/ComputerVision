# Track and Predict with YOLOv11

import required libraries
- YOLO - Ultralytics YOLOv11 model for object detection and tracking

```python
from ultralytics import YOLO
```

create a YOLO model instance using the pre-trained weights from "yolo11n.pt"

```python
model = YOLO("../models/yolo11n.pt")
```

Available trackers:
- ByteTrack - A high-performance multi-object tracker ( bytetrack.yaml )
- BoT-SORT - A robust tracker for various scenarios ( botsort.yaml )

```python
model.track(
    persist=True,
    source="../media/video.mov",
    conf=0.6,  # 60% confidence score
    line_width=4,
    device="cpu",
    save=True,
    tracker="bytetrack.yaml"
)
```

```python
model = YOLO("../models/yolo11n.pt")
stream_video = 'https://youtu.be/TsNe0edwWcs?si=TDt0gXEmKqruS6FA' # use YouTube as source

max_frames = 300 # limit the max number of frames

for i, result in enumerate(
        model.track(
            source=stream_video,      # use youtube video as source
            persist=True,             # keep tracking results
            tracker="bytetrack.yaml", # use ByteTrack as tracker
            stream=True,              # yields frame-by-frame results
            verbose=False,            # supress sys-out
            save=True,                # save output
        )
):
    if i + 1 >= max_frames:
        print(f"Reached max frames limit of {max_frames}. Stopping tracking.")
        break
```
