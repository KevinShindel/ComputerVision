# Object Counting Solution
- using the ObjectCounting module from the Ultralytics solutions package, we can easily count objects in a video stream. The module provides a simple interface to detect and count objects based on a pre-trained model. In this example, we will use a YOLOv8 model to count objects in a video file.

import the necessary libraries and initialize the video capture and writer objects. We will also define the model to be used for object counting.

```python
import cv2
from ultralytics import solutions
```

initialize the video capture and writer objects. We will read the input video file and get its width, height, and frames per second (fps). Then, we will create a VideoWriter object to save the output video with the counted objects.

```python
cap = cv2.VideoCapture("../media/bottles_video.mov")
w, h, fps = (
    int(cap.get(x))
    for x in (cv2.CAP_PROP_FRAME_WIDTH,
              cv2.CAP_PROP_FRAME_HEIGHT,
              cv2.CAP_PROP_FPS))

video_writer = (
    cv2.VideoWriter(
        "result.avi",
        cv2.VideoWriter_fourcc(*"mp4v"),
        fps, (w, h)))
```

create an instance of the ObjectCounter class from the Ultralytics solutions package.

 We will specify the model to be used, the classes to be counted (if any), and other parameters such as whether to show the input and output frames, the region of interest for counting, and the line width for drawing bounding boxes.

```python
# ObjectCounting Module Initialization
model = 'best-detect.pt'
solution = solutions.ObjectCounter(
    model=model,
    # classes=[39],
    show_in=True,
    show_out=True,
    region=[(640, 0), (640, 720)],
    line_width=4,
)


# Process video
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        break
    results = solution(im0)
    video_writer.write(results.plot_im)
cap.release()
video_writer.release()
```
