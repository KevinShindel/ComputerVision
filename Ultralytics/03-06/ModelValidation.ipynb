# YOLOv8

```python
from ultralytics import YOLO
import json
```

load model

```python
model = YOLO("best-segment.pt")
```

model validation

```python
metrics = model.val(
    data="data.yaml"
)
```

metrics extraction

```python
metrics_dict = metrics.results_dict
json_dump = json.dumps(metrics_dict, indent=4, sort_keys=True)
print(json_dump)

# save metrics to a json file
with open("metrics.json", "w") as f:
    f.write(json_dump)
```
