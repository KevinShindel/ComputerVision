# TrackZone Solution
- When need to track objects in a specific zone, you can use the TrackZone module.
- The module will detect and track objects in the specified zone and ignore objects outside the zone.

import required libraries
- cv2 - OpenCV library for image and video processing
- ultralytics - Ultralytics library for object detection and tracking

```python
import cv2
from ultralytics import solutions
```

create a video writer to save the output video

```python
# read video from local file
cap = cv2.VideoCapture("../04-02/video.mp4")

# collect width, height and frame-rate from video
w, h, fps = (
    int(cap.get(x))
    for x in (cv2.CAP_PROP_FRAME_WIDTH,
              cv2.CAP_PROP_FRAME_HEIGHT,
              cv2.CAP_PROP_FPS))

# create video writer
video_writer = (
    cv2.VideoWriter(
        "result.avi",
        cv2.VideoWriter_fourcc(*"mp4v"),
        fps, (w, h)))
```

create a TrackZone solution with the best model and set the line width to 3

```python
# TrackZone Module Initialization
solution = solutions.TrackZone(
    model="../04-02/best-detect.pt", # use pre-trained model
    line_width=3 # set width for track zone to 3px
)
```

inference model and save the output video

```python
# Process video
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        break
    results = solution(im0)
    video_writer.write(results.plot_im)
cap.release()
video_writer.release()
```
