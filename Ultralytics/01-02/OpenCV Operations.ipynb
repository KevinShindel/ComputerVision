# Basic Image Operations

import required libraries
- **cv2** - for image processing
- **matplotlib** - for visualization
- **IPython** - for video visualization in Jupyter Notebooks

```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
```

```python
# Operation-1: Read the image from directory
image = cv2.imread("../media/apples_image.jpg", cv2.IMREAD_COLOR_RGB)

# Operation-2: Blur the image
blurred = cv2.blur(image, (50, 50))
```

```python
plt.imshow(image)
```

```python
plt.imshow(blurred)
```

```python
cv2.imwrite('blurred.jpg', blurred)
```

```python
x1, x2 = 240, 680
y1, y2 = 450, 900
sliced_img = image[x1:x2, y1:y2]

plt.imshow(sliced_img)
```

```python
cap = cv2.VideoCapture("../media/apples_video.mov")

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    clear_output(wait=True)
    plt.figure(figsize=(8, 4))
    plt.imshow(frame_rgb)
    plt.axis("off")
    display(plt.gcf())
    plt.close()

cap.release()
```
