# Auto-annotate images in the dataset

- [Annotator Docs](https://docs.ultralytics.com/reference/data/annotator#ultralytics.data.annotator.auto_annotate)

#### import libraries

- **os** - for interacting with the operating system, such as file and directory management.
- **shutil** - for high-level file operations, such as copying and moving files.
- **auto_annotate** - a function from the ultralytics.data.annotator module that automatically annotates images for object detection tasks.

```python
import os
import shutil

from ultralytics.data.annotator import auto_annotate
```

## Auto-annotate images in the dataset

- Using detection and segmentation models to automatically annotate images in the specified dataset directories.

```python
paths = (
    "dataset/train/images",
    "dataset/val/images",
    "dataset/test/images"
)

for p in paths:
    print(f'Auto-annotating {p}')
    auto_annotate(data=p,
                  det_model="yolo26s.pt",  # detection model
                  sam_model="sam_b.pt") # segmentation model
    print(f'Auto-annotation completed for {p}')
```

## Remove old labels and rename auto-annotated labels to "labels" folder

```python
for dp, dn, _ in os.walk("dataset"):
    has_auto = "images_auto_annotate_labels" in dn
    has_labels = "segments" in dn

    if not has_auto:
        continue  # keep existing segments if no new auto-annotated segments here

    labels_path = os.path.join(dp, "segments")
    auto_labels_path = os.path.join(dp, "images_auto_annotate_labels")

    if has_labels:
        print(f"Removing folder: {labels_path}")
        shutil.rmtree(labels_path)

    print(f"Renaming folder: {auto_labels_path} to {labels_path}")
    os.rename(auto_labels_path, labels_path)
```
