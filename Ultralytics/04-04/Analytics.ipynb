# Ultralytics Analytics Module
- turns object detection and tracking results into real-time charts, so you can watch how object counts change across a video frame by frame

import required libraries
- **cv2** - OpenCV library for video processing
- **ultralytics** - Ultralytics library for object detection and tracking

```python
import cv2
from ultralytics import solutions
```

create a video capture object to read the input video file and get its properties such as width, height, and frames per second (fps). Then, create a video writer object to save the processed video with analytics overlays.

```python
cap = cv2.VideoCapture("../media/vegetables_video.mov")
w, h, fps = (
    int(cap.get(x))
    for x in (cv2.CAP_PROP_FRAME_WIDTH,
              cv2.CAP_PROP_FRAME_HEIGHT,
              cv2.CAP_PROP_FPS))
video_writer = (
    cv2.VideoWriter(
        "result.avi",
        cv2.VideoWriter_fourcc(*"mp4v"),
        fps, (1280, 720)))
```

Analytics Module Initialization

```python
solution = solutions.Analytics(
    model="new-model.pt",
    analytics_type="bar",
    device="cpu",
)
```

Process video

```python
frame_count = 0

while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        break
    frame_count += 1
    results = solution(im0, frame_count)
    video_writer.write(results.plot_im)
cap.release()
video_writer.release()
```
