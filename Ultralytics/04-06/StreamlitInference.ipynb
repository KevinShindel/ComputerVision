# Streamlit Inference

- [Streamlit Docs](https://docs.ultralytics.com/guides/streamlit-live-inference#introduction)

Under the hood, the Streamlit application uses the Ultralytics solutions module to create an interactive interface. When you start the inference, the application:

1. Captures video from your webcam or uploaded video file
2. Processes each frame through the YOLO26 model
3. Applies object detection with your specified confidence and IoU thresholds
4. Displays both the original and annotated frames in real-time
5. Optionally enables object tracking if selected

The application provides a clean, user-friendly interface with controls to adjust model parameters and start/stop inference at any time.

```python
from ultralytics import solutions
```

create an instance of the Inference class, specifying the model you want to use for inference. You can choose any model supported by Ultralytics, such as YOLO11 or YOLOv10. The model file should be in the same directory or provide the full path to the model file.

```python
inf = solutions.Inference(
    model="new-model.pt",  # you can use any model that Ultralytics support, i.e. YOLO11, YOLOv10
)
```

run the inference method to start the Streamlit application. This will launch a web interface where you can interact with the model, upload videos, and see the results in real-time. Make sure to run this script in an environment where Streamlit is installed and accessible.

```python
inf.inference()
```

#### Make sure to run the file using command `streamlit run path/to/file.py`
