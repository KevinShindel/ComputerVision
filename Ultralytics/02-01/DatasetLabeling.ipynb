# Dataset Labeling

Preparation Steps:

1. [Install Label Studio](https://labelstud.io/guide/install#Install-using-pip): `pip install label-studio`
2. set maximum files limit to 10000 `set DATA_UPLOAD_MAX_NUMBER_FILES=10000` via Command shell !
2. [Start Label Studio](https://labelstud.io/guide/start): `label-studio start`
3. [Create a new project](https://labelstud.io/guide/setup_project) in Label Studio and configure the labeling interface according to your dataset requirements.
4. [Upload your dataset](https://labelstud.io/guide/tasks) to Label Studio for labeling.
5. [Start labeling](https://labelstud.io/guide/labeling) your dataset using the configured interface.
